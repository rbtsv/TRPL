```python
%load_ext memory_profiler 
# %memit

max1_tests_path = "./1-HelloWorld/1-2-max/tests"
cont_lis_path, cont_lis_tst = "../1-HelloWorld/cont_lis/", "t4"

def create_input(tests_path=max1_tests_path, test_num = "01"):
    test_file = open(f"{tests_path}/{test_num}", "r")
    return test_file.readline

input = create_input()

def read_int():
    return int(input())

def print_cf_style_input(tests_path=max1_tests_path, test_num = "01"):
    test_file = open(f"{tests_path}/{test_num}", "r")
    print("входные данные:")
    print(test_file.read())
    print("выходные данные:\n")
```

```python
class InputIterator:
    def __init__(self, tests_path=max1_tests_path, test_num = "01"):
        self.input = create_input(tests_path, test_num)
        self.n = int(self.input())
        self.cur = 0
        
    def __iter__(self):        
        return self
    
    def __next__(self):
        if self.cur < self.n:
            self.cur += 1
            return(int(self.input()))
        raise StopIteration
        
```

## Числа Фибоначчи

```python
def Fib(n):
    if n <= 2:
        return 1
    return Fib(n-1) + Fib(n-2)
```

```python
[Fib(i) for i in range(1,10)]
```

```python
def iter_Fib(n):
    a, b = 1, 1
    for i in range(n-2):
        a, b = a+b, a
    return a
```

```python
[iter_Fib(i) for i in range(1,10)]
```

```python
%%time
Fib(40)
```

```python
%%time
iter_Fib(40)
```

```python
from functools import lru_cache

@lru_cache
def lru_Fib(n):
    if n <= 2:
        return 1
    return lru_Fib(n-1) + lru_Fib(n-2)
```

```python
%%time
lru_Fib(40)
```

```python
%%time
lru_Fib(1000)
```

```python
%%time
iter_Fib(1000)
```

```python
def Fib(n):          # 0
    if n <= 2:       
        r = 1
    else:        
        a = Fib(n-1) # 1
        b = Fib(n-2) # 2
        r = a + b    
    return r         # 3
```

```python
def iter_stack_Fib(x):
    s = [(0, x, None, None, None)] # (line, n, a, b, r)
    while s:
        line, n, a, b, r = s.pop()
        if line == 3:
            result = r
        elif line == 0:
            if n <= 2:       
                #result = 1
                s.append((3, n, a, b, 1))
            else:      
                s.append((1, n, a, b, r))
                s.append((0, n-1, None, None, None))
        elif line == 1:
            a = result
            s.append((2, n, a, b, r))
            s.append((0, n-2, None, None, None))
        elif line == 2:
            b = result
            s.append((3, n, a, b, a+b))            
    return result
```

```python
[iter_stack_Fib(i) for i in range(1,10)]
```

```python
%%time
iter_stack_Fib(40)
```

```python
%%time
Fib(40)
```

```python
def iter_stack_Fib(x):
    s = [(0, x, None, None)] # (line, n, a, b)
    while s:
        line, n, a, b = s.pop()        
        if line == 0:
            if n <= 2:       
                result = 1                
            else:      
                s.append((1, n, a, b))
                s.append((0, n-1, None, None))
        elif line == 1:
            a = result
            s.append((2, n, a, b))
            s.append((0, n-2, None, None))
        elif line == 2:
            b = result
            result = a+b
    return result
```

```python
[iter_stack_Fib(i) for i in range(1,10)]
```
